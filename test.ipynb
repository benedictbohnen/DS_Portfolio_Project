```python
import pandas as pd

pfad = "data/en.openfoodfacts.org.products.csv.gz"   # an deinen Dateinamen anpassen

header = pd.read_csv(pfad, sep="\t", nrows=0)   # nur die Kopfzeile lesen
print(len(header.columns), "Spalten")

gesucht = [
    "code", "product_name", "countries_tags", "categories_tags",
    "ingredients_analysis_tags", "nutriscore_grade", "nova_group",
    "energy-kcal_100g", "fat_100g", "saturated-fat_100g",
    "sugars_100g", "salt_100g", "proteins_100g", "fiber_100g",
]
cols = [c for c in gesucht if c in header.columns]
fehlend = [c for c in gesucht if c not in header.columns]
print("vorhanden:", cols)
print("fehlt:", fehlend)
```

```python
teile = []
gelesen = 0

for chunk in pd.read_csv(
    pfad,
    sep="\t",
    usecols=cols,
    chunksize=200_000,
    low_memory=False,
    on_bad_lines="skip",
):
    gelesen += len(chunk)
    de = chunk[chunk["countries_tags"].str.contains("en:germany", na=False)]
    teile.append(de)
    print(f"{gelesen:,} Zeilen gelesen, bisher {sum(len(t) for t in teile):,} aus Deutschland")

df = pd.concat(teile, ignore_index=True)
df.to_csv("data/produkte_de.csv", index=False)
print(df.shape)
```
