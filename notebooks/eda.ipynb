```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Anteile in Prozent, Reihenfolge der Gruppen festlegen
anteile = pd.crosstab(df["diet_type"], df["nova_group"], normalize="index") * 100
anteile = anteile.loc[["Vegan", "Vegetarisch", "Nicht-vegetarisch"]]

# 2. Eine Farbe pro NOVA-Gruppe (grün = gering verarbeitet, rot = ultra-verarbeitet)
nova_farben = ["#2e8b57", "#f4c542", "#f08a24", "#c0392b"]   # NOVA 1, 2, 3, 4

# 3. Säulendiagramm mit gruppierten Säulen
ax = anteile.plot(
    kind="bar",
    figsize=(10, 5),
    color=nova_farben,
    width=0.8,
)

# 4. Prozentwerte über die Säulen schreiben
for container in ax.containers:
    ax.bar_label(container, fmt="%.0f%%", fontsize=9, padding=2)

# 5. Beschriftung
ax.set_title("Verarbeitungsgrad (NOVA) nach Ernährungsform", fontsize=13, weight="bold")
ax.set_xlabel("")
ax.set_ylabel("Anteil der Produkte (%)")
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
ax.legend(title="NOVA-Gruppe", frameon=False)
sns.despine()
plt.tight_layout()
plt.show()
```
